```python


import numpy as np
from PIL import Image
from math import sqrt

import torch
import torch.nn as nn
import torch.nn.functional as F
import matplotlib.pyplot as plt

device = torch.device('cuda:0' if torch.cuda.is_available() else 'cpu')
print('PyTorch:', torch.__version__, '/ device:', device)
if device.type == 'cpu':
    print('※ GPU が使われていません．「ランタイムのタイプを変更」で GPU を選択してください．')



class Equalized_LR:
    def __init__(self, name):
        self.name = name

    def compute_weight(self, module):
        weight = getattr(module, self.name+'_orig')
        fan_in = weight.data.size(1) * weight.data[0][0].numel()
        return weight * sqrt(2 / fan_in)

    @staticmethod
    def apply(module, name):
        fn = Equalized_LR(name)

        weight = getattr(module, name)
        del module._parameters[name]
        module.register_parameter(name + '_orig', nn.Parameter(weight.data))
        module.register_forward_pre_hook(fn)

        return fn

    def __call__(self, module, input):
        weight = self.compute_weight(module)
        setattr(module, self.name, weight)

def equal_lr(module, name='weight'):
    Equalized_LR.apply(module, name)
    return module


class EqualConv2d(nn.Module):
    def __init__(self, *args, **kwargs):
        super().__init__()

        conv = nn.Conv2d(*args, **kwargs)
        conv.weight.data.normal_()
        conv.bias.data.zero_()
        self.conv = equal_lr(conv)

    def forward(self, input):
        return self.conv(input)


class EqualLinear(nn.Module):
    def __init__(self, in_dim, out_dim, mapping_linear=False,
                 gain=None, lrmul=None, bias=True, use_wscale=False):
        super().__init__()

        self.mapping_linear = mapping_linear
        linear = nn.Linear(in_dim, out_dim)
        linear.weight.data.normal_()
        linear.bias.data.zero_()

        if mapping_linear:
            he_std = gain * in_dim ** (-0.5)
            if use_wscale:
                init_std = 1.0 / lrmul
                self.w_mul = he_std * lrmul
            else:
                init_std = he_std / lrmul
                self.w_mul = lrmul
            self.weight = torch.nn.Parameter(torch.randn(out_dim, in_dim) * init_std)
            if bias:
                self.bias = torch.nn.Parameter(torch.zeros(out_dim))
                self.b_mul = lrmul
            else:
                self.bias = None
        else:
            self.linear = equal_lr(linear)

    def forward(self, input):
        if self.mapping_linear:
            bias = self.bias
            if bias is not None:
                bias = bias * self.b_mul
            return F.linear(input, self.weight * self.w_mul, bias)
        else:
            return self.linear(input)


class FusedUpsample(nn.Module):
    def __init__(self, in_channel, out_channel, kernel_size, padding=0):
        super().__init__()

        weight = torch.randn(in_channel, out_channel, kernel_size, kernel_size)
        bias = torch.zeros(out_channel)

        fan_in = in_channel * kernel_size * kernel_size
        self.multiplier = sqrt(2 / fan_in)

        self.weight = nn.Parameter(weight)
        self.bias = nn.Parameter(bias)

        self.pad = padding

    def forward(self, input):
        weight = F.pad(self.weight * self.multiplier, [1, 1, 1, 1])
        weight = (
            weight[:, :, 1:, 1:]
            + weight[:, :, :-1, 1:]
            + weight[:, :, 1:, :-1]
            + weight[:, :, :-1, :-1]
        ) / 4

        out = F.conv_transpose2d(input, weight, self.bias, stride=2, padding=self.pad)

        return out

class FusedDownsample(nn.Module):
    def __init__(self, in_channel, out_channel, kernel_size, padding=0):
        super().__init__()

        weight = torch.randn(out_channel, in_channel, kernel_size, kernel_size)
        bias = torch.zeros(out_channel)

        fan_in = in_channel * kernel_size * kernel_size
        self.multiplier = sqrt(2 / fan_in)

        self.weight = nn.Parameter(weight)
        self.bias = nn.Parameter(bias)

        self.pad = padding

    def forward(self, input):
        weight = F.pad(self.weight * self.multiplier, [1, 1, 1, 1])
        weight = (
            weight[:, :, 1:, 1:]
            + weight[:, :, :-1, 1:]
            + weight[:, :, 1:, :-1]
            + weight[:, :, :-1, :-1]
        ) / 4

        out = F.conv2d(input, weight, self.bias, stride=2, padding=self.pad)

        return out



class PixelNorm(nn.Module):
    '''
    Pixel-wise normalization
    '''
    def __init__(self):
        super().__init__()

    def forward(self, input):
        return input / torch.sqrt(torch.mean(input ** 2, dim=1, keepdim=True) + 1e-8)

class AdaptiveInstanceNorm(nn.Module):
    '''
    Adaptive instance normalization
    '''
    def __init__(self, in_channel, style_dim):
        super().__init__()

        self.norm = nn.InstanceNorm2d(in_channel)
        self.style = EqualLinear(style_dim, in_channel * 2)

        self.style.linear.bias.data[:in_channel] = 1
        self.style.linear.bias.data[in_channel:] = 0

    def forward(self, input, style):
        style = self.style(style).unsqueeze(2).unsqueeze(3)
        gamma, beta = style.chunk(2, 1)

        out = self.norm(input)
        out = gamma * out + beta

        return out



torch.manual_seed(0)
adain = AdaptiveInstanceNorm(in_channel=4, style_dim=8)
x = torch.randn(1, 4, 16, 16) * 5 + 3           # 平均3・標準偏差5の特徴マップ
w = torch.randn(1, 8)                           # スタイルコード
out = adain(x, w)

gamma, beta = adain.style(w).chunk(2, 1)
print('入力のチャネル平均   :', x.mean(dim=[2, 3]).squeeze().numpy().round(3))
print('出力のチャネル平均   :', out.mean(dim=[2, 3]).squeeze().detach().numpy().round(3))
print('beta                :', beta.squeeze().detach().numpy().round(3))
print('出力のチャネル標準偏差:', out.std(dim=[2, 3], unbiased=False).squeeze().detach().numpy().round(3))
print('|gamma|             :', gamma.abs().squeeze().detach().numpy().round(3))

"""### その他のレイヤ
* ConstantInput：Generatorに入力する任意のサイズの定数マップを作成するためのクラス．
* NoiseInjection：GeneratorのAdaINの前に与えるノイズを作成するために必要なクラス．
* Blur：ダウンサンプル，アップサンプル前後でローパスフィルタを施すために必要となるクラス．
"""

class ConstantInput(nn.Module):
    def __init__(self, channel, size=4):
        super().__init__()

        self.input = nn.Parameter(torch.randn(1, channel, size, size))

    def forward(self, batch):
        out = self.input.repeat(batch, 1, 1, 1)
        return out

class NoiseInjection(nn.Module):
    def __init__(self, channel):
        super().__init__()
        self.weight = nn.Parameter(torch.zeros(1, channel, 1, 1))

    def forward(self, image, noise):
        return image + self.weight * noise

class Blur(nn.Module):
    def __init__(self, kernel=None, normalize=None, flip=None, stride=1):
        super(Blur, self).__init__()
        self.normalize = normalize
        self.flip = flip
        if kernel is None:
            kernel = [1, 2, 1]
        kernel = self.create_kernel(kernel)

        self.register_buffer('kernel', kernel)
        self.stride = stride


    def create_kernel(self, kernel):
        kernel = torch.tensor(kernel).float().view(1, len(kernel))
        kernel = kernel.t() * kernel
        kernel = kernel.unsqueeze(0).unsqueeze(0)

        if self.normalize:
            kernel /= kernel.sum()
        if self.flip:
            kernel = torch.flip(kernel, [2, 3])
        return kernel

    def forward(self, input):
        kernel = self.kernel.expand(input.size(1), -1, -1, -1)
        padding_size = int((self.kernel.size(2) - 1) / 2)
        return F.conv2d(input, kernel, stride=self.stride,
                        padding=padding_size, groups=input.size(1))



class GBlock(nn.Module):
    def __init__(self, in_features, out_features, kernel_size, stride, padding,
                 up=False, fused=False, n_style=512, initial=False):
        super(GBlock, self).__init__()
        if not initial:
            if up:
                if fused:
                    self.c1 = nn.Sequential(
                        FusedUpsample(in_features, out_features, kernel_size=kernel_size, padding=padding),
                        Blur([1, 2, 1], True, True, 1))
                else:
                    self.c1 = nn.Sequential(
                        nn.Upsample(scale_factor=2, mode='bilinear', align_corners=False),
                        Blur([1, 2, 1], True, True, 1),
                        EqualConv2d(in_features, out_features, kernel_size=kernel_size, stride=stride,
                                    padding=padding, bias=True)
                    )
            else:
                self.c1 = EqualConv2d(in_features, out_features, kernel_size=kernel_size, stride=stride,
                                          padding=padding, bias=True)

        self.initial = initial
        self.make_noise1 = NoiseInjection(out_features)
        self.adain1 = AdaptiveInstanceNorm(out_features, n_style)
        self.lrelu1 = nn.LeakyReLU(negative_slope=0.2)

        self.c2 = nn.Sequential(
            EqualConv2d(out_features, out_features, kernel_size=kernel_size, stride=stride,
                        padding=padding, bias=True))
        self.make_noise2 = NoiseInjection(out_features)
        self.adain2 = AdaptiveInstanceNorm(out_features, n_style)
        self.lrelu2 = nn.LeakyReLU(negative_slope=0.2)

    def forward(self, x, style_code, noise):
        out = x
        if not self.initial:
            out = self.c1(out)
        out = self.make_noise1(out, noise)
        out = self.lrelu1(out)
        out = self.adain1(out, style_code)

        out = self.c2(out)
        out = self.make_noise2(out, noise)
        out = self.lrelu2(out)
        out = self.adain2(out, style_code)
        return out

class ConvBlock(nn.Module):
    def __init__(self, in_features, out_features, kernel_size, stride,
                 padding, down=False, fused=False, d_last=False):
        super(ConvBlock, self).__init__()

        conv_layer1 = [
            EqualConv2d(in_features, out_features, kernel_size=kernel_size, stride=stride,
                        padding=padding, bias=True),
            nn.LeakyReLU(negative_slope=0.2)
        ]

        if d_last:
            kernel_size = 4
            stride = 1
            padding = 0

        if down:
            if fused:
                conv_layer2 = [
                    Blur([1, 2, 1], True, True, 1),
                    FusedDownsample(out_features, out_features, kernel_size=kernel_size, padding=padding),
                    nn.LeakyReLU(negative_slope=0.2)
                ]
            else:
                conv_layer2 = [
                    Blur([1, 2, 1], True, True, 1),
                    EqualConv2d(out_features, out_features, kernel_size=kernel_size, stride=stride,
                        padding=padding, bias=True),
                    nn.AvgPool2d(2, 2),
                    nn.LeakyReLU(negative_slope=0.2)
                ]
        else:
            conv_layer2 = [
                EqualConv2d(out_features, out_features, kernel_size=kernel_size, stride=stride,
                            padding=padding, bias=True),
                nn.LeakyReLU(negative_slope=0.2)
            ]

        self.c1 = nn.Sequential(*conv_layer1)
        self.c2 = nn.Sequential(*conv_layer2)

    def forward(self, x):
        h = self.c1(x)
        out = self.c2(h)
        return out


class Generator(nn.Module):
    def __init__(self, switch_ch=4, img_size=1024, n_ch=512, n_style=512, n_mapping_layers=8):
        super(Generator, self).__init__()
        n_block = int(np.log2(img_size))

        module_list = []
        conv_module_list = []
        fused = False
        in_ch = n_ch
        out_ch = n_ch
        for block_idx in range(1, n_block+1):
            if block_idx == 1:initial = True
            else:initial = False
            module_list.append(GBlock(in_ch, out_ch, kernel_size=3, stride=1, padding=1, up=True, fused=fused,
                                      n_style=n_style, initial=initial))
            conv_module_list.append(ConvBlock(out_ch, 3, kernel_size=1, stride=1, padding=0))
            if block_idx >= switch_ch:
                fused = True
                in_ch = out_ch
                out_ch = in_ch // 2

        mapping_network = [PixelNorm()]
        for i in range(n_mapping_layers):
            mapping_network.append(EqualLinear(n_style, n_style, mapping_linear=True,
                                   gain=2 ** 0.5, lrmul=0.01, bias=True, use_wscale=True))
            mapping_network.append(nn.LeakyReLU(negative_slope=0.2))

        self.const = ConstantInput(n_ch)
        self.main_net = nn.ModuleList(module_list)
        self.to_rgb = nn.ModuleList(conv_module_list)
        self.mapping_network = nn.Sequential(*mapping_network)

    def forward(self, style, noise, depth=0, alpha=-1, true_size=1024):
        const = self.const(style.size(0))
        style_code = self.mapping_network(style)

        if depth > 0 and alpha < 1.0:
            h = const
            presize_noise = torch.randn(h.size(0), 1, h.size(2), h.size(3)).type_as(style)
            for i in range(depth - 1):
                size = true_size // (2 ** (depth - i))
                noise_roop = torch.randn(h.size(0), 1, size, size).type_as(style)
                h = self.main_net[i](h, style_code, noise_roop)
                presize_noise = torch.randn(h.size(0), 1, h.size(2)*2, h.size(3)*2).type_as(style)

            h1 = self.main_net[depth - 1](h, style_code, presize_noise)
            h2 = F.interpolate(h1, scale_factor=2, mode='bilinear', align_corners=False)
            h3 = self.to_rgb[depth - 1](h2)
            h4 = self.main_net[depth](h1, style_code, noise)
            h4 = self.to_rgb[depth](h4)

            h = h3 - alpha * (h3 - h4)
        else:
            h = const
            for i in range(depth):
                size = true_size // (2 ** (depth - i))
                noise_roop = torch.randn(h.size(0), 1, size, size).type_as(style)
                h = self.main_net[i](h, style_code, noise_roop)

            h = self.main_net[depth](h, style_code, noise)
            h = self.to_rgb[depth](h)

        return h



class Discriminator(nn.Module):
    def __init__(self, switch_ch=4, img_size=1024, n_ch=16):
        super(Discriminator, self).__init__()
        n_block = int(np.log2(img_size))
        self.total_depth = n_block - 1

        module_list = []
        conv_module_list = []
        down = True
        fused = False
        d_last = False
        in_ch = n_ch
        out_ch = n_ch
        for block_idx in range(1, n_block+1):
            if block_idx == n_block:
                down = False
                d_last = True
            conv_module_list.append(nn.Sequential(EqualConv2d(3, in_ch, kernel_size=1, stride=1, padding=0),
                                    nn.LeakyReLU(negative_slope=0.2)))
            module_list.append(ConvBlock(in_ch, out_ch, kernel_size=3, stride=1, padding=1, down=down, fused=fused, d_last=d_last))
            if block_idx <= switch_ch + 1:
                fused = True
                in_ch = out_ch
                out_ch = in_ch * 2
            else:
                in_ch = out_ch

        self.main_net = nn.Sequential(*module_list)
        self.from_rgb = nn.Sequential(*conv_module_list)
        self.linear = EqualLinear(out_ch, 1)

    def forward(self, x, depth=0, alpha=0):
        if depth > 0 and alpha < 1.0:
            h1 = self.from_rgb[self.total_depth - depth](x)
            h1 = self.main_net[self.total_depth - depth](h1)
            x2 = F.avg_pool2d(x, 2, 2)
            h2 = F.leaky_relu(self.from_rgb[self.total_depth - depth+1](x2))
            h = h2 - alpha * (h2 - h1)

        else:
            h = self.from_rgb[self.total_depth - depth](x)
            h = self.main_net[self.total_depth - depth](h)

        for i in range(depth):
            h = self.main_net[self.total_depth - depth+1+i](h)

        out = self.linear(h.view(h.size(0), -1))
        return out


torch.manual_seed(0)
G_toy = Generator(switch_ch=4, img_size=64, n_ch=64, n_style=512, n_mapping_layers=8).to(device)
D_toy = Discriminator(switch_ch=4, img_size=64, n_ch=16).to(device)
print('Generator のパラメータ数    : {:,}'.format(sum(p.numel() for p in G_toy.parameters())))
print('Discriminator のパラメータ数: {:,}'.format(sum(p.numel() for p in D_toy.parameters())))
print()

z = torch.randn(2, 512, device=device)
with torch.no_grad():
    for d in range(0, 6):
        size = 4 * 2 ** d
        noise = torch.randn(2, 1, size, size, device=device)
        for alpha in ([1.0] if d == 0 else [0.5, 1.0]):
            img = G_toy(z, noise, depth=d, alpha=alpha, true_size=size)
            score = D_toy(img, depth=d, alpha=alpha)
            print(f'depth={d}, alpha={alpha:.1f} : G出力 {tuple(img.shape)} → D出力 {tuple(score.shape)}')



import os, sys, re

REPO = '/content/stylegan3' if os.path.exists('/content') else os.path.abspath('stylegan3')
if not os.path.exists(REPO):
    !git clone -q https://github.com/NVlabs/stylegan3.git {REPO}

# --- 新しい Python / setuptools 向けの修正 ---------------------------------
def _patch(path, old, new):
    path = os.path.join(REPO, path)
    src = open(path, encoding='utf-8').read()
    if old in src:
        open(path, 'w', encoding='utf-8').write(src.replace(old, new))

_patch('dnnlib/util.py', 'from distutils.util import strtobool',
       "strtobool = lambda v: v.lower() in ('y', 'yes', 't', 'true', 'on', '1')")
for f in ['torch_utils/ops/conv2d_gradfix.py', 'torch_utils/ops/grid_sample_gradfix.py']:
    _patch(f, 'from pkg_resources import parse_version', 'from packaging.version import parse as parse_version')
# -----------------------------------------------------------------------------

if REPO not in sys.path:
    sys.path.insert(0, REPO)
import dnnlib, legacy
from torch_utils.ops import bias_act, upfirdn2d
from training import networks_stylegan2

USE_CUSTOM_CUDA_OPS = False  #@param {type:"boolean"}
if not USE_CUSTOM_CUDA_OPS:
    bias_act._init = lambda: False     # PyTorch による参照実装を使う
    upfirdn2d._init = lambda: False
print('準備完了')




def my_modulated_conv2d(x, weight, styles, demodulate=True, padding=1):
    '''
    x      : 入力特徴マップ [N, Cin, H, W]
    weight : 畳み込みの重み [Cout, Cin, k, k]
    styles : スタイル s     [N, Cin]   （w からアフィン変換で求めたもの）
    '''
    N, Cin, H, W = x.shape
    Cout, _, k, _ = weight.shape

    # (1) modulation : 入力チャネルごとにスタイルを掛ける    w' = s_i * w
    w = weight[None] * styles[:, None, :, None, None]           # [N, Cout, Cin, k, k]

    # (2) demodulation : 出力チャネルごとに正規化            w'' = w' / sqrt(sum w'^2)
    if demodulate:
        d = (w.square().sum(dim=[2, 3, 4]) + 1e-8).rsqrt()      # [N, Cout]
        w = w * d[:, :, None, None, None]

    # (3) グループ畳み込みで，サンプルごとに異なる重みで畳み込む
    x = x.reshape(1, N * Cin, H, W)
    w = w.reshape(N * Cout, Cin, k, k)
    out = F.conv2d(x, w, padding=padding, groups=N)
    return out.reshape(N, Cout, H, W)


torch.manual_seed(0)
x = torch.randn(4, 16, 8, 8, device=device)
weight = torch.randn(32, 16, 3, 3, device=device)
styles = torch.randn(4, 16, device=device)

out_mine = my_modulated_conv2d(x, weight, styles)
out_official = networks_stylegan2.modulated_conv2d(x=x, weight=weight, styles=styles, padding=1)
print('出力サイズ        :', tuple(out_mine.shape))
print('公式実装との最大誤差:', (out_mine - out_official).abs().max().item())
print('出力の標準偏差     :', out_mine.std().item(), '（demodulation により約1になる）')


src = open(networks_stylegan2.__file__, encoding='utf-8').read()
start = src.index('def modulated_conv2d(')
print(src[start:src.index('#----', start)])



MODEL = 'stylegan2-celebahq-256x256.pkl'  #@param ['stylegan2-celebahq-256x256.pkl', 'stylegan2-ffhq-256x256.pkl', 'stylegan2-ffhq-512x512.pkl', 'stylegan2-ffhq-1024x1024.pkl', 'stylegan2-afhqv2-512x512.pkl', 'stylegan2-afhqcat-512x512.pkl', 'stylegan2-afhqdog-512x512.pkl', 'stylegan2-metfaces-1024x1024.pkl', 'stylegan2-lsundog-256x256.pkl']

url = 'https://api.ngc.nvidia.com/v2/models/nvidia/research/stylegan2/versions/1/files/' + MODEL
with dnnlib.util.open_url(url) as f:
    G = legacy.load_network_pkl(f)['G_ema'].to(device).eval()   # G_ema：重みを移動平均したGenerator
for p in G.parameters():
    p.requires_grad_(False)

print(f'z の次元 : {G.z_dim}')
print(f'w の次元 : {G.w_dim}')
print(f'出力解像度: {G.img_resolution}')
print(f'num_ws   : {G.num_ws}  （synthesis network の各層に1つずつ与える w の数）')



print(G.mapping)
print()

# 各解像度のブロックがどの w を使うか
layer_of_w = [[] for _ in range(G.num_ws)]
w_idx = 0
for res in G.synthesis.block_resolutions:
    block = getattr(G.synthesis, f'b{res}')
    convs = ['conv1'] if block.num_conv == 1 else ['conv0', 'conv1']
    for j, name in enumerate(convs):
        layer_of_w[w_idx + j].append(f'{res}x{res}.{name}')
    layer_of_w[w_idx + block.num_conv].append(f'{res}x{res}.torgb')
    w_idx += block.num_conv

for i, layers in enumerate(layer_of_w):
    print(f'w[{i:2d}] → ' + ', '.join(layers))

"""### 画像表示用の関数"""

def to_numpy_img(img):
    '''[N,3,H,W] の値域[-1,1]のテンソル → [N,H,W,3] の uint8 配列'''
    return (img.permute(0, 2, 3, 1) * 127.5 + 128).clamp(0, 255).to(torch.uint8).cpu().numpy()

def seed_to_z(seed):
    return torch.from_numpy(np.random.RandomState(seed).randn(1, G.z_dim)).float().to(device)

def show_grid(imgs, ncol, titles=None, row_labels=None, size=2.2):
    '''imgs: [N,H,W,3] の配列のリスト'''
    imgs = list(imgs)
    nrow = (len(imgs) + ncol - 1) // ncol
    fig, axes = plt.subplots(nrow, ncol, figsize=(size * ncol, size * nrow), squeeze=False)
    for i, ax in enumerate(axes.flat):
        ax.set_xticks([]); ax.set_yticks([])
        for s in ax.spines.values(): s.set_visible(False)
        if i < len(imgs):
            ax.imshow(imgs[i])
            if titles is not None: ax.set_title(titles[i], fontsize=9)
    if row_labels is not None:
        for r, lab in enumerate(row_labels):
            axes[r, 0].set_ylabel(lab, fontsize=10)
    plt.tight_layout()
    plt.show()

@torch.no_grad()
def synth(ws, noise_mode='const'):
    return to_numpy_img(G.synthesis(ws, noise_mode=noise_mode))


seeds = [0, 1, 2, 3, 4, 5, 6, 7]
truncation_psi = 0.7  #@param {type:"slider", min:0, max:1.2, step:0.05}

imgs = []
with torch.no_grad():
    for s in seeds:
        z = seed_to_z(s)
        w = G.mapping(z, None, truncation_psi=truncation_psi)   # [1, num_ws, w_dim]
        imgs.append(synth(w)[0])
show_grid(imgs, ncol=4, titles=[f'seed={s}' for s in seeds], size=3)



psis = [0.0, 0.3, 0.5, 0.7, 1.0, 1.2]
seeds = [10, 11, 12]
imgs = []
with torch.no_grad():
    for s in seeds:
        z = seed_to_z(s)
        for psi in psis:
            imgs.append(synth(G.mapping(z, None, truncation_psi=psi))[0])
show_grid(imgs, ncol=len(psis), titles=[f'ψ={p}' for p in psis] * len(seeds))



seed_a, seed_b = 20, 21  #@param
n_interp = 8
psi = 0.7

z_a, z_b = seed_to_z(seed_a), seed_to_z(seed_b)
ts = np.linspace(0, 1, n_interp)
imgs_z, imgs_w = [], []
with torch.no_grad():
    w_a = G.mapping(z_a, None, truncation_psi=psi)
    w_b = G.mapping(z_b, None, truncation_psi=psi)
    for t in ts:
        z = (1 - t) * z_a + t * z_b                     # Z 空間で補間
        imgs_z.append(synth(G.mapping(z, None, truncation_psi=psi))[0])
        w = (1 - t) * w_a + t * w_b                     # W 空間で補間
        imgs_w.append(synth(w)[0])
show_grid(imgs_z + imgs_w, ncol=n_interp, titles=[f't={t:.2f}' for t in ts] * 2,
          row_labels=['Z で補間', 'W で補間'], size=1.8)


from IPython.display import HTML
import matplotlib.animation as animation

plt.rcParams['animation.embed_limit'] = 60   # MB
n_frames = 20
frames = []
with torch.no_grad():
    for t in np.linspace(0, 1, n_frames):
        im = synth((1 - t) * w_a + t * w_b)[0]
        frames.append(np.asarray(Image.fromarray(im).resize((256, 256), Image.LANCZOS)))
frames = frames + frames[::-1]

fig = plt.figure(figsize=(4, 4))
plt.axis('off')
ims = [[plt.imshow(im, animated=True)] for im in frames]
ani = animation.ArtistAnimation(fig, ims, interval=80, repeat_delay=500, blit=True)
plt.close()
HTML(ani.to_jshtml())

## スタイルミキシング


src_seeds = [30, 31, 32]       # 画像A（行）
dst_seeds = [40, 41, 42, 43]   # 画像B（列）
psi = 0.7


mix_ranges = {
    'Coarse (4〜8)':  range(0, 4),
    'Middle (16〜32)': range(4, 8),
    'Fine (64〜)':    range(8, G.num_ws),
}

with torch.no_grad():
    ws_src = [G.mapping(seed_to_z(s), None, truncation_psi=psi) for s in src_seeds]
    ws_dst = [G.mapping(seed_to_z(s), None, truncation_psi=psi) for s in dst_seeds]

    for name, idx in mix_ranges.items():
        idx = list(idx)
        imgs = [np.full((G.img_resolution, G.img_resolution, 3), 255, np.uint8)]  # 左上は空白
        imgs += [synth(w)[0] for w in ws_dst]
        for w_a in ws_src:
            imgs.append(synth(w_a)[0])
            for w_b in ws_dst:
                w_mix = w_a.clone()
                w_mix[:, idx] = w_b[:, idx]      # 指定した層の w だけを B のものに入れ替える
                imgs.append(synth(w_mix)[0])
        print(f'■ {name} の層の w を B から借りた場合 （w[{idx[0]}]〜w[{idx[-1]}]）')
        show_grid(imgs, ncol=len(dst_seeds) + 1, size=1.8,
                  titles=[''] + [f'B: seed={s}' for s in dst_seeds] + [''] * (len(imgs) - len(dst_seeds) - 1),
                  row_labels=[''] + [f'A: seed={s}' for s in src_seeds])



seed_a, seed_b = 50, 51  #@param
n_interp = 7
psi = 0.7

with torch.no_grad():
    w_a = G.mapping(seed_to_z(seed_a), None, truncation_psi=psi)
    w_b = G.mapping(seed_to_z(seed_b), None, truncation_psi=psi)
    imgs, labels = [], []
    for name, idx in list(mix_ranges.items()) + [('すべての層', range(G.num_ws))]:
        idx = list(idx)
        for t in np.linspace(0, 1, n_interp):
            w = w_a.clone()
            w[:, idx] = (1 - t) * w_a[:, idx] + t * w_b[:, idx]
            imgs.append(synth(w)[0])
        labels.append(name)
show_grid(imgs, ncol=n_interp, row_labels=labels, size=1.8,
          titles=[f't={t:.2f}' for t in np.linspace(0, 1, n_interp)] + [''] * (len(imgs) - n_interp))



seed = 60  #@param
with torch.no_grad():
    w = G.mapping(seed_to_z(seed), None, truncation_psi=0.7)
    imgs = [synth(w, noise_mode='random')[0] for _ in range(4)]
    imgs.append(synth(w, noise_mode='none')[0])

    many = torch.cat([G.synthesis(w, noise_mode='random') for _ in range(100)])
    std = many.std(dim=0).mean(dim=0).cpu().numpy()
    std_img = (plt.cm.inferno(std / max(std.max(), 1e-8))[:, :, :3] * 255).astype(np.uint8)

show_grid(imgs + [std_img], ncol=6, size=2.5,
          titles=['ノイズ1', 'ノイズ2', 'ノイズ3', 'ノイズ4', 'ノイズなし', '標準偏差'])

# 髪の部分を拡大して比較
c = G.img_resolution
crop = (slice(0, c // 3), slice(c // 3, 2 * c // 3))
show_grid([im[crop] for im in imgs[:4]], ncol=4, size=3,
          titles=['ノイズ1（拡大）', 'ノイズ2（拡大）', 'ノイズ3（拡大）', 'ノイズ4（拡大）'])

```
